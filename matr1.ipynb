```python
import numpy as np

sizes = [100, 200, 400, 800, 1200, 1600]

for n in sizes:
    A = np.random.rand(n, n)
    B = np.random.rand(n, n)

    with open(f'matrix_A_{n}.txt', 'w') as f:
        f.write(f"{n}\n")
        for i in range(n):
            for j in range(n):
                f.write(f"{A[i][j]:.6f} ")
            f.write("\n")

    with open(f'matrix_B_{n}.txt', 'w') as f:
        f.write(f"{n}\n")
        for i in range(n):
            for j in range(n):
                f.write(f"{B[i][j]:.6f} ")
            f.write("\n")

    print(f"Созданы матрицы {n}x{n}")
```

```python
import numpy as np

with open("result_matrix.txt") as f:
    n = int(f.readline())
    data = f.read()
    C = np.array([list(map(float, line.split())) for line in data.strip().split('\n')])

with open(f"matrix_A_{n}.txt") as f:
    f.readline()
    data = f.read()
    A = np.array([list(map(float, line.split())) for line in data.strip().split('\n')])

with open(f"matrix_B_{n}.txt") as f:
    f.readline()
    data = f.read()
    B = np.array([list(map(float, line.split())) for line in data.strip().split('\n')])


diff = np.max(np.abs(A @ B - C))
print(f"Размер: {n}")
print("Верификация успешна" if diff < 1e-3 else "Ошибка")
print(f"Макс. разница: {diff:.2e}")
```
